```python
from pyannote.database.util import load_rttm
from pyannote.core import notebook, Segment
import json


# Load Diarization from DetectAudio.py
with open('results_whisperx_diarization.json') as f:
    dia = json.load(f)

# Mount RTTM output
speakersRTTM = ""
i = 0
for x, y in dia["segment"].items():
    speakersRTTM = speakersRTTM + "SPEAKER resultsCardiacEscenarioVideo 1 " + str(y["start"]) + " " + str(y["duration"]) + " <NA> <NA> " + dia["speaker"][str(x)] + " <NA> <NA>\n"
    i = i + 1

# Write rttm into file
f = open("resultsCardiacEscenarioVideo.rttm", "w")
f.write(speakersRTTM)
f.close()

# Read RTTM file & Show Speakers
REFERENCE_SCENARIO = "resultsCardiacEscenarioVideo.rttm"
rttmEscenario = load_rttm(REFERENCE_SCENARIO)["resultsCardiacEscenarioVideo"]

notebook.crop = Segment(0, 580)
rttmEscenario

```
